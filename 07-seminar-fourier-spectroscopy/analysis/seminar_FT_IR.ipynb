```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

# —————————————————————————————————————————————————————————————————————————————
# CONFIGURATION & STYLING
# —————————————————————————————————————————————————————————————————————————————
plt.rcParams.update({
    'font.family': 'serif',
    'font.size': 11,
    'axes.labelsize': 12,
    'axes.titlesize': 12,
    'xtick.labelsize': 10,
    'ytick.labelsize': 10,
    'legend.fontsize': 10
})

# Wir erstellen 3 Subplots untereinander, die sich die Achsen teilen
fig, (ax1, ax2, ax3) = plt.subplots(3, 1, figsize=(7.5, 8.5), sharex=True, sharey=True, dpi=300)

# Daten laden
df_0 = pd.read_csv("ft_ir_none.csv", comment="#")
df_2x = pd.read_csv("ft_ir_2x_zeropadding.csv", comment="#")
df_8x = pd.read_csv("ft_ir_8x_zeropadding.csv", comment="#")
df_32x = pd.read_csv("ft_ir_32x_zeropadding.csv", comment="#")

df_triangle_2 = pd.read_csv("ft_ir_triangle_2x_zeropadding.csv", comment="#")
df_triangle_8 = pd.read_csv("ft_ir_triangle_8x_zeropadding.csv", comment="#")
df_triangle_32 = pd.read_csv("ft_ir_triangle_32x_zeropadding.csv", comment="#")

# —————————————————————————————————————————————————————————————————————————————
# SUBPLOT 1: 1x vs. 2x (Erste Füllung der Lücken)
# —————————————————————————————————————————————————————————————————————————————
ax1.plot(df_0["Wavelength"], df_0["CH1 FFT"], color='black', lw=1.5, label='Ohne Zero-Padding')
ax1.scatter(df_0["Wavelength"], df_0["CH1 FFT"], color='black', s=20, zorder=5)

ax1.plot(df_2x["Wavelength"], df_2x["CH1 FFT"], color='#4daf4a', lw=1.2, label='2x Zero-Padding')
ax1.plot(df_triangle_2["Wavelength"], df_triangle_2["CH1 FFT"], color='purple', lw=1.3, ls='--', label='2x Zero-Padding + Triangle')

ax1.legend(loc='lower right', frameon=True, edgecolor='none', framealpha=0.8)
ax1.grid(True, linestyle=':', alpha=0.5)

# —————————————————————————————————————————————————————————————————————————————
# SUBPLOT 2: 1x vs. 8x (Deutliche Annäherung an die Linienform)
# —————————————————————————————————————————————————————————————————————————————
ax2.plot(df_0["Wavelength"], df_0["CH1 FFT"], color='black', lw=1.5, label='Ohne Zero-Padding')
ax2.scatter(df_0["Wavelength"], df_0["CH1 FFT"], color='black', s=20, zorder=5)

ax2.plot(df_8x["Wavelength"], df_8x["CH1 FFT"], color='#984ea3', lw=1.2, label='8x Zero-Padding')
ax2.plot(df_triangle_8["Wavelength"], df_triangle_8["CH1 FFT"], color='darkorange', lw=1.3, ls='--', label='8x Zero-Padding + Triangle')

ax2.legend(loc='lower right', frameon=True, edgecolor='none', framealpha=0.8)
ax2.grid(True, linestyle=':', alpha=0.5)
ax2.set_ylabel(r'Intensität $I$ / $\mathrm{a.u.}$')

# —————————————————————————————————————————————————————————————————————————————
# SUBPLOT 3: 1x vs. 32x (Perfekt glatte, kontinuierliche Sinc-Interpolation)
# —————————————————————————————————————————————————————————————————————————————
ax3.plot(df_0["Wavelength"], df_0["CH1 FFT"], color='black', lw=1.5, label='Ohne Zero-Padding')
ax3.scatter(df_0["Wavelength"], df_0["CH1 FFT"], color='black', s=20, zorder=5)

ax3.plot(df_32x["Wavelength"], df_32x["CH1 FFT"], color='#e41a1c', lw=1.2, label='32x Zero-Padding')
ax3.plot(df_triangle_32["Wavelength"], df_triangle_32["CH1 FFT"], color='darkcyan', lw=1.3, ls='--', label='32x Zero-Padding + Triangle')

ax3.legend(loc='lower right', frameon=True, edgecolor='none', framealpha=0.8)
ax3.grid(True, linestyle=':', alpha=0.5)
ax3.set_xlabel(r'Wellenlänge $\lambda$ / $\mathrm{nm}$')

# Grenzen festlegen
plt.xlim(1846, 1850)
plt.ylim(20, 80)

# Diskreter Rahmen für alle Subplots
for ax in [ax1, ax2, ax3]:
    ax.spines['top'].set_alpha(0.3)
    ax.spines['right'].set_alpha(0.3)

fig.suptitle('Zusammenspiel von Interpolationsdichte (Zero-Padding) und Apodisation', y=0.98, fontsize=13)
plt.tight_layout()
plt.savefig('Zero_Padding_Triangle_Subplots.png', bbox_inches='tight', dpi=300)
plt.show()
```

```python


# bestees ergebnis
df_0 = pd.read_csv("ft_ir_triangle_32x_zeropadding.csv", comment="#")
wellenlänge_0 = df_0["Wavelength"].values
wellenzahl_0 = df_0["Wavenumber"].values
amplitude_0 = df_0["CH1 FFT"].values
amplitude_ref_0 = df_0["Reference FFT"].values

plt.rcParams.update({
    'font.family': 'serif',
    'font.size': 11,
    'axes.labelsize': 12,
    'axes.titlesize': 12,
    'xtick.labelsize': 10,
    'ytick.labelsize': 10,
    'legend.fontsize': 10,
    'figure.titlesize': 14
})

# Erstellung der Figur im typischen einspaltigen Paper-Format (ca. 8.5 cm Breite)
fig, ax = plt.subplots(figsize=(6, 4.5))
#fig, ax = plt.subplots(figsize=(6, 4.5), dpi=300)

# Plots zeichnen
# Referenz zuerst, damit die Hauptmessung physisch darüber liegt

ax.plot(wellenlänge_0, amplitude_ref_0,color='#e41a1c', alpha=0.3, lw=1.2, ls='--', 
        label='Referenzspektrum')

ax.plot(wellenlänge_0, amplitude_0, color='darkslateblue', lw=1.5, 
        label='Natrium-Spektrum ($d = 22\,\mathrm{mm}$')

# —————————————————————————————————————————————————————————————————————————————
# 3. ACHSEN, TITEL & LEGENDE
# —————————————————————————————————————————————————————————————————————————————
# Präzise Achsenbeschriftung mit korrekten physikalischen Einheiten in LaTeX
ax.set_xlabel(r'Wellenlänge $\lambda$ / $\mathrm{nm}$')
ax.set_ylabel(r'Intensität $I$ / $10^3\,\mathrm{a.u.}$')
ax.set_title('Emissionsspektrum mittels FT-IR', pad=12)

# Grenzen exakt wie vorgegeben (angepasst an die Skalierung / 1000)
ax.set_ylim(0, 80)

# Dezentes, professionelles Grid
ax.grid(True, which='both', linestyle=':', color='gray', alpha=0.5)

# Rahmen (Spines) oben und rechts sichtbar lassen, aber dezent (Standard im Physik-Paper)
ax.spines['top'].set_visible(True)
ax.spines['right'].set_visible(True)
ax.spines['top'].set_alpha(0.3)
ax.spines['right'].set_alpha(0.3)

# Legende ohne störenden dicken Rahmen, leicht transparent platziert
ax.legend(loc='upper left', frameon=True, framealpha=0.9, edgecolor='none')

# Layout optimieren und speichern
plt.tight_layout()
fig.savefig('FT_IR.pdf', bbox_inches='tight') # Vektorgrafik für LaTeX-Dokumente
```
