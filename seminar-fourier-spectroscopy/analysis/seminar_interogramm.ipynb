```python
natrium = np.genfromtxt(
    'Interferogramm_Na_korr.txt',
    delimiter='\t',
    converters={
        0: lambda s: float(s.decode('utf-8').replace(',', '.')),
        1: lambda s: float(s.decode('utf-8').replace(',', '.'))
    }
)

# X-Achse: Von nm oder mm in µm umrechnen 
spiegelverschiebung_na = natrium[:, 0] / 1000  
amplitude_na = natrium[:, 1]

plt.rcParams.update({
    'font.family': 'serif',
    'font.size': 11,
    'axes.labelsize': 12,
    'axes.titlesize': 12,
    'xtick.labelsize': 10,
    'ytick.labelsize': 10,
    'legend.fontsize': 10,
    'figure.titlesize': 14
})

fig, ax = plt.subplots(figsize=(6, 4.5)) # dpi=300 wieder rein für scharfe Vorschau

# Plot zeichnen mit lw=0.5 (sehr gut für dichte Oszillationen!)
ax.plot(spiegelverschiebung_na, amplitude_na, 
        color='plum', lw=0.5,
        label='Interferogramm') # LABEL KORRIGIERT

# —————————————————————————————————————————————————————————————————————————————
# 3. ACHSEN, TITEL & LEGENDE
# —————————————————————————————————————————————————————————————————————————————
# Wissenschaftlich korrekte Achsenbeschriftung mit Schrägstrich
ax.set_xlabel(r'Spiegelverschiebung $x$ / $\mathrm{\mu m}$')
ax.set_ylabel(r'Intensität $I$ / $\mathrm{a.u.}$')
ax.set_title('Interferogramm der Natrium-Dampflampe', pad=12) # TITEL VERVOLLSTÄNDIGT

# Grenzen der X-Achse (prüfen, ob die Schwebung komplett im Fenster 0-8 µm liegt)
ax.set_xlim(0, 8) 

# Dezentes, professionelles Grid
ax.grid(True, which='both', linestyle=':', color='gray', alpha=0.5)

# Rahmen-Styling
ax.spines['top'].set_visible(True)
ax.spines['right'].set_visible(True)
ax.spines['top'].set_alpha(0.3)
ax.spines['right'].set_alpha(0.3)

# Legende
ax.legend(loc='upper right', frameon=True, framealpha=0.9, edgecolor='none')

plt.tight_layout()

# Speicherbefehle für morgen (einfach Auskommentierung löschen bei Bedarf)
fig.savefig('Natrium_Interferogramm.pdf', bbox_inches='tight')

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# —————————————————————————————————————————————————————————————————————————————
# 1. DATEN EINLESEN & SKALIEREN
# —————————————————————————————————————————————————————————————————————————————
weißlicht = np.genfromtxt(
    'Interferogramm_WL_korr.txt',
    delimiter='\t',
    converters={
        0: lambda s: float(s.decode('utf-8').replace(',', '.')),
        1: lambda s: float(s.decode('utf-8').replace(',', '.'))
    }
)

# X-Achse: Skalierung auf µm (prüfen, ob /1000 für die Rohdaten passt)
spiegelverschiebung_weiß = weißlicht[:, 0] / 1000
amplitude_weiß = weißlicht[:, 1]

# —————————————————————————————————————————————————————————————————————————————
# 2. PLOT-STYLING (PUBLICATION READY)
# —————————————————————————————————————————————————————————————————————————————
plt.rcParams.update({
    'font.family': 'serif',
    'font.size': 11,
    'axes.labelsize': 12,
    'axes.titlesize': 12,
    'xtick.labelsize': 10,
    'ytick.labelsize': 10,
    'legend.fontsize': 10,
    'figure.titlesize': 14
})

# dpi=300 aktivieren für gestochen scharfe Vektorschau in Jupyter
fig, ax = plt.subplots(figsize=(6, 4.5))

# Plot zeichnen – Farbe passend zur Glühlampe (z.B. warmes Orange oder ein edles Grau)
ax.plot(spiegelverschiebung_weiß, amplitude_weiß, 
        color='plum',
        label='Interferogramm (Glühbirne)') # LABEL KORRIGIERT

# —————————————————————————————————————————————————————————————————————————————
# 3. ACHSEN, TITEL & LEGENDE
# —————————————————————————————————————————————————————————————————————————————
# Wissenschaftlich korrekte Achsenbeschriftung mit Schrägstrich (DIN 1313)
ax.set_xlabel(r'Spiegelverschiebung $x$ /$\mathrm{\mu m}$')
ax.set_ylabel(r'Intensität $I$ / $\mathrm{a.u.}$')
ax.set_title('Interferogramm der Weißlichtquelle (Glühlampe)', pad=12)

# Grenzen anpassen: Wichtig! Der Centerburst muss genau in der Mitte liegen.
# Wenn der Centerburst z.B. bei 4 µm liegt, wähle ax.set_xlim(2, 6)
ax.set_xlim(0, 8) 

# Dezentes, professionelles Grid
ax.grid(True, which='both', linestyle=':', color='gray', alpha=0.5)

# Rahmen dezent einblenden
ax.spines['top'].set_visible(True)
ax.spines['right'].set_visible(True)
ax.spines['top'].set_alpha(0.3)
ax.spines['right'].set_alpha(0.3)

# Legende ohne störenden dicken Rahmen
ax.legend(loc='upper right', frameon=True, framealpha=0.9, edgecolor='none')

plt.tight_layout()

# Export-Befehle für Folien/Protokoll:
fig.savefig('Weisslicht_Interferogramm.pdf', bbox_inches='tight')

plt.show()
```
